```python
#two pointers and sliding window
class Solution:
    def LSC(self, s:str) -> int:
        last = {}
        left = 0
        best = 0
        for right, ch in unumerate(s):
            if ch in last and last[ch] >= left:
                left = last[ch] + 1
            last[ch] = right
            best = max(best, right - left + 1)
        return best
```
