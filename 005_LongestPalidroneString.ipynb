```python
class Solution:
    def longestPalidrome(self, s: str) -> str:
        t = '#' + '#'.join(s) + '#'
        max = 0
        start = 0

        for i in range(len(t)):
            l = r = i
            while l >= 0 and r < len(t) and t[l] == t[r]:
                length = (r - l) // 2

                if length > max:
                    max = length
                    start = l // 2
            l -= 1
            r += 1
        return s[start : start + max]            
```
