```python
class Solution:
    def MedianTwoArray(self, l1: List[int], l2: List[int]) -> float:
        res = l1[:]
        for i in len(l2):
            res.append(l2[i])

        a = sorted(res)
        if len(a) % 2 == 0:
            return (a[len(a) // 2] + a[len(a)//2 - 1]) / 2
        else : 
            return a[len((a) // 2)]    
```
