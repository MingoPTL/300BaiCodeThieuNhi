```python
class Solution:
    def ktra(self, s: str) -> bool:
        stack = []
        ngoac = {")" : "(","]" : "[","}" : "{"}

        for c in s:
            if c in ngoac:
                if not stack or stack[-1] != ngoac[c]:
                    return False
                stack.pop()
            else:
                stack.append(c)
        return len(stack) == 0
```
