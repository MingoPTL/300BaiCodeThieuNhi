```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
```

```python

class Solution:
    def removeNthFromEnd(self, head: ListNode | None, n: int) -> ListNode | None:
        cur = head
        pos = 0
        while cur is not None:
            cur =  cur.next
            pos+=1

        step = 0
        tmp = ListNode(0)
        tmp.next = head
        prev = tmp

        while step < pos - n:
            prev = prev.next
            step+=1
        
        prev.next = prev.next.next
        return tmp.next
```
