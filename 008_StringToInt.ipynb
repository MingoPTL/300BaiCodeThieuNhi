```python
class Solution:
    def myAtoi(self, s: str) -> int:
        s = s.lstrip()
        sign = 1
        i = 0
        res = 0

        if not s:
            return 0

        if s[i] == '-':
            sign = -1
            i+=1
        elif s[i] == "+":
            i+=1
        
        while i < len(s):
            if '0' <= s[i] <= '9':
                digit = int(s[i])
                res = res * 10 + digit
                i+=1
            else : 
                break
        res = res * sign
        if res < -(2**31):
            return -(2**31)
        elif res > (2**31) - 1:
            return (2**31) - 1
        return res
```
