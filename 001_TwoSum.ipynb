```python
class Solution:
    def twoSum(self, nums, target):
        a = sorted((v, i) for i, v in enumerate(nums))

        l, r = 0, len(a) - 1
        while l < r:
            s = a[l][0] + a[r][0]
            if s == target:
                return [a[l][1], a[r][1]]
            elif s < target:
                l += 1
            else:
                r -= 1
        return []
```
