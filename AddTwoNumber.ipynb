```python
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        tmp = ListNode(0)
        current = tmp
        sonho = 0

        while l1 or l2 or sonho:
            s = sonho
            if l1:
                s+= l1.val
                l1= l1.next
            if l2:
                s+= l2.val
                l2= l2.next 
            sonho = s//10
            current.next = ListNode(s%10)
            current = current.next
        return tmp.next
```
