```python
class Solution:
    def conmbinePhone(self, digits : str) -> list[str]:
        if not digits:
            return []
        
        mapping = {
            "2": "abc", "3": "def", "4": "ghi", "5": "jkl",
            "6": "mno", "7": "pqrs", "8": "tuv", "9": "wxyz",
        }

        res = [""]
        for d in digits:
            tmp = []
            for s in res:
                for ch in mapping[d]:
                    tmp.append(s+ch)
            res = tmp
        return res

```
