```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def mergeTwoLists(self, list1: ListNode | None, list2: ListNode | None) -> ListNode | None:
        kq =  ListNode(0)
        tmp = kq
        head1 = list1
        head2 = list2

        while head1 and head2:
            if head1.val <= head2.val:
                tmp.next = head1
                head1 = head1.next
            elif head2.val <= head1.val:
                tmp.next = head2
                head2 = head2.next
            tmp = tmp.next
        if head1:
            tmp.next = head1
        if head2:
            tmp.next = head2
        return kq.next
```
