```python
class Solution:
    def removeDuplicate(self, nums; List[int])-> int:
        left, right = 0, 1
        while right < len(nums):
            if nums[right] == nums[left]:
                right+=1
            elif nums[right] != nums[left]:
                left+=1
                nums[left] = nums[right]
        return left + 1
```
