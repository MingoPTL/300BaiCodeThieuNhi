```python
class Solution:
    def ReverseInt(self, x: int) -> int:
        res = 0
        if x > 0 :
            while x > 0 : 
                sonho = x % 10
                x = x // 10
                res = res * 10 + sonho
            if res < -(2**31) or res > (2**31) - 1 : 
                return 0
            return res    

        elif x < 0:
            tmp = x * -1
            while tmp > 0 : 
                sonho = tmp % 10
                tmp = tmp // 10
                res = res * 10 + sonho
            if res < -(2**31) or res > (2**31) - 1 : 
                return 0
            return res * -1
        else :
            return 0


```
